Open this notebook → turn on **GPU** → **Run the cell below** → open the **Cloudflare link** it prints.

Clones official [`unslothai/unsloth`](https://github.com/unslothai/unsloth). Studio UI is **AGPL-3.0** — keep this credit if you reshare.


```python
# @title ▶ Run once — install + start Unsloth Studio
import os, sys, subprocess
from pathlib import Path

def detect_root() -> Path:
    if Path("/content").exists():
        return Path("/content")
    if Path("/kaggle/working").exists():
        return Path("/kaggle/working")
    return Path.cwd()

ROOT = detect_root()
REPO = ROOT / "unsloth"
print(f"Root: {ROOT}")

need_clone = not (REPO / "studio" / "setup.sh").exists()
if need_clone:
    if REPO.exists():
        subprocess.run(["rm", "-rf", str(REPO)], check=False)
    print("Cloning unslothai/unsloth ...")
    subprocess.run(
        ["git", "clone", "--depth", "1", "--branch", "main",
         "https://github.com/unslothai/unsloth.git", str(REPO)],
        check=True,
    )
else:
    print(f"Using existing repo: {REPO}")

os.chdir(REPO)
subprocess.run(["chmod", "+x", "studio/setup.sh"], check=True)
print("Running setup.sh --local (few minutes on first run)...")
r = subprocess.run(["./studio/setup.sh", "--local"])
if r.returncode != 0:
    raise SystemExit(
        "Setup failed. Reset runtime, enable GPU + Internet, run this cell again."
    )

backend = REPO / "studio" / "backend"
if not (backend / "colab.py").exists():
    raise FileNotFoundError("Setup finished but colab.py missing — reset and retry.")

print("")
print("✅ Setup complete — starting Studio...")
print("➡️  Open the Cloudflare URL below in a NEW tab (iframe often blank).")
print("")

sys.path.insert(0, str(backend))
from colab import start
start()
```
