# Unsloth Studio on Kaggle

Run open AI models in your browser for free — **chat** or light **fine-tuning** — using [Unsloth Studio](https://unsloth.ai/docs/new/studio).

### Before you start (first time on Kaggle)
1. Go to [kaggle.com](https://www.kaggle.com) and **sign up / sign in**
2. **Verify your phone** when Kaggle asks (do this first — needed for GPU)
3. Complete any quick account checks so GPU access unlocks
4. Open **this notebook**, then click **Copy and Edit** (you need your own copy to run it)
5. Turn on GPU + Internet under **Session options** (not the top-menu Settings):
   - On the **right panel** (Input / Output / files), expand it if needed (small arrow near the bottom-right)
   - Open **Session options**
   - **Accelerator** → **GPU T4 x2** (or P100)
   - **Internet** / **Enable Internet** → **On**
6. You’re ready for the steps below

### How to use
1. Confirm GPU + Internet are on under **Session options**
2. **Run the cell below** (press ▶)

**First run usually takes about 4–5 minutes** (download + install). That’s normal. When it’s done you should see a **Shareable Unsloth Link** card with a Cloudflare URL and a **password**.

### After it starts
1. Click **Open Unsloth Studio** (or open the Cloudflare URL in a **new tab**)
2. Enter the **password** from the card (log in as `unsloth`)
   - Only if the card did **not** show a password, run this in a new cell: `!cat /root/.unsloth/studio/auth/.bootstrap_password`
3. Load a model (see recommendations below)
4. Leave the install/start cell **running** while you use Studio

### Recommended models on free Kaggle (GPU T4×2 ≈ lots of VRAM)
Unsloth supports 500+ open models — these are **strong and practical** on this free GPU (prefer **GGUF / Q4** or similar quants in Model Hub):

**Best everyday picks**
- **Qwen3.5 4B / 9B** — excellent quality-per-size; great default for chat + coding
- **Gemma 4 E4B / 12B** — strong Google open models; E4B is light, 12B is sharper
- **Llama 3.2 3B** or **Llama 3.1 8B** — reliable, widely used baselines

**Stronger (still usually fine on T4×2 with Q4 GGUF)**
- **Qwen3 14B** — very capable general / reasoning chat
- **Phi-4 14B** — strong for its size
- **gpt-oss 20B** — solid open option Unsloth ships notebooks for
- **Gemma 4 26B-A4B** — MoE-style efficiency; good if you want bigger quality without full 31B cost

**Start here if unsure:** Model Hub → **Qwen3.5 9B** (Q4 GGUF) or type `hi` in Chat for a small default.

Skip giant dense 70B+ on free Kaggle — they’ll OOM or crawl.

### Good to know
- Free Kaggle GPU is about **~30 hours/week**, sessions up to ~**12 hours**
- You do **not** need to fine-tune — chatting alone is fine
- This notebook only launches the official Unsloth installer ([`unslothai/unsloth`](https://github.com/unslothai/unsloth)). Studio UI is AGPL-3.0.


```python
# @title ▶ Run this cell — installs and starts Unsloth Studio (~4–5 min first run)
import os, sys, subprocess
from pathlib import Path

ROOT = Path("/kaggle/working")
REPO = ROOT / "unsloth"
print(f"Working in: {ROOT}")
print("First run usually takes about 4–5 minutes. Please wait...")
print("")

need_clone = not (REPO / "studio" / "setup.sh").exists()
if need_clone:
    if REPO.exists():
        subprocess.run(["rm", "-rf", str(REPO)], check=False)
    print("Downloading Unsloth (first run only)...")
    subprocess.run(
        ["git", "clone", "--depth", "1", "--branch", "main",
         "https://github.com/unslothai/unsloth.git", str(REPO)],
        check=True,
    )
else:
    print("Unsloth already downloaded — reusing it.")

os.chdir(REPO)
subprocess.run(["chmod", "+x", "studio/setup.sh"], check=True)
print("Installing Unsloth Studio...")
r = subprocess.run(["./studio/setup.sh", "--local"])
if r.returncode != 0:
    raise SystemExit(
        "Install failed. Restart the session, enable GPU + Internet under Session options, then run again."
    )

backend = REPO / "studio" / "backend"
if not (backend / "colab.py").exists():
    raise FileNotFoundError("Install finished but Studio files are missing — reset and retry.")

print("")
print("✅ Install done — starting Studio...")
print("")
print("Next steps:")
print("  1) Use the Shareable Unsloth Link card (Open Unsloth Studio / Cloudflare URL)")
print("  2) Log in as unsloth with the password on the card")
print("  3) If no password on the card: !cat /root/.unsloth/studio/auth/.bootstrap_password")
print("  4) Model Hub: try Qwen3.5 9B or Gemma 4 12B (Q4) — or Chat and type hi")
print("  5) Keep THIS cell running — stopping it shuts Studio down")
print("")

sys.path.insert(0, str(backend))
from colab import start
start()
```

### Stuck?
- **Phone / account verification** — finish Kaggle phone verify first; GPU options stay greyed out until then
- **Can’t find Internet / GPU** — right panel → expand → **Session options** (top menu Settings won’t show these)
- **Still installing** — wait the full ~4–5 minutes on first run
- **Password already on the card** — use that; no need to `cat` the file
- **No password on the card** — `!cat /root/.unsloth/studio/auth/.bootstrap_password`
- **Install failed** — Factory reset session → Session options GPU + Internet → run once
