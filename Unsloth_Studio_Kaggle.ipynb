# Unsloth Studio on Kaggle

Run open AI models in your browser for free — **chat** or light **fine-tuning** — using [Unsloth Studio](https://unsloth.ai/docs/new/studio).

### How to use (about 2 steps)
1. Right sidebar → **Settings** → Accelerator **GPU T4 x2** (or P100) → **Internet: On**
2. **Run the cell below** (press ▶) and wait until install finishes

### After it starts — read this
The cell will print:
- a **Cloudflare URL** (something like `https://….trycloudflare.com`)
- an **admin username** and **password**

**Do this:**
1. Open the Cloudflare URL in a **new tab**
2. Log in with the username + password shown in the cell output
3. Go to **Chat**, pick a small model (e.g. Qwen3.5 4B or Llama 3.2 3B), download it, and send a message
4. Leave this Kaggle cell **running** while you use Studio

### Good to know
- Free Kaggle GPU is about **~30 hours/week**, sessions up to ~**12 hours**
- T4×2 usually has more VRAM than free Colab — nicer for mid-size models
- You do **not** need to fine-tune — chatting alone is fine
- This notebook only launches the official Unsloth installer ([`unslothai/unsloth`](https://github.com/unslothai/unsloth)). Studio UI is AGPL-3.0.


```python
# @title ▶ Run this cell — installs and starts Unsloth Studio
import os, sys, subprocess
from pathlib import Path

ROOT = Path("/kaggle/working")
REPO = ROOT / "unsloth"
print(f"Working in: {ROOT}")

need_clone = not (REPO / "studio" / "setup.sh").exists()
if need_clone:
    if REPO.exists():
        subprocess.run(["rm", "-rf", str(REPO)], check=False)
    print("Downloading Unsloth (first run only)...")
    subprocess.run(
        ["git", "clone", "--depth", "1", "--branch", "main",
         "https://github.com/unslothai/unsloth.git", str(REPO)],
        check=True,
    )
else:
    print("Unsloth already downloaded — reusing it.")

os.chdir(REPO)
subprocess.run(["chmod", "+x", "studio/setup.sh"], check=True)
print("Installing Unsloth Studio (a few minutes on first run)...")
r = subprocess.run(["./studio/setup.sh", "--local"])
if r.returncode != 0:
    raise SystemExit(
        "Install failed. Reset the runtime, make sure GPU is on, then run this cell again."
    )

backend = REPO / "studio" / "backend"
if not (backend / "colab.py").exists():
    raise FileNotFoundError("Install finished but Studio files are missing — reset and retry.")

print("")
print("✅ Install done — starting Studio...")
print("")
print("When you see a Cloudflare link below:")
print("  1) Copy the URL and open it in a NEW browser tab")
print("  2) Copy the admin username + password printed in this output")
print("  3) Log in on that page (password is shown only here)")
print("  4) In Studio → Chat → pick a small model and try a message")
print("  5) Keep THIS cell running — stopping it shuts Studio down")
print("")

sys.path.insert(0, str(backend))
from colab import start
start()
```

### Stuck?
- **No Cloudflare link** — scroll the cell output; wait for install to fully finish
- **Internet / GPU errors** — confirm Settings: GPU on + Internet On, then restart session and re-run
- **Forgot the password** — it’s only in the cell output; restart session and run again for a new login
- **Install failed** — Factory reset session, enable GPU + Internet, run the cell once
