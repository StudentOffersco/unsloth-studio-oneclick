# Unsloth Studio on Google Colab

Run open AI models in your browser for free — **chat** or light **fine-tuning** — using [Unsloth Studio](https://unsloth.ai/docs/new/studio).

### How to use
1. **Runtime → Change runtime type → GPU (T4) → Save**
2. **Run the cell below** (press ▶)

**First run usually takes about 4–5 minutes** (download + install). That’s normal. When it’s done you’ll see a **Shareable Unsloth Link** card with a Cloudflare URL and a **password**.

### After it starts
1. Click **Open Unsloth Studio** (or open the Cloudflare URL in a **new tab**)
2. Enter the **password** shown on the card (the card says log in as `unsloth` — you mainly just need that password)

![Login — password only](https://raw.githubusercontent.com/StudentOffersco/unsloth-studio-oneclick/main/images/01-login-password.png)

3. Load a model (see recommendations below)
4. Leave this Colab cell **running** while you use Studio

### Recommended models on free Colab (GPU T4 ≈ 15GB VRAM)
Unsloth supports 500+ open models — these are **comfortable** on free Colab (prefer **GGUF / Q4** in Model Hub):

**Best everyday picks**
- **Qwen3.5 4B** — great default for chat + light coding
- **Gemma 4 E4B** — strong and light
- **Llama 3.2 3B** — fast, reliable baseline
- **Qwen3.5 0.8B / 2B** — if you want something tiny and snappy

**Still usually fine (Q4 GGUF; may be tighter)**
- **Qwen3.5 9B**
- **Llama 3.1 8B**
- **Gemma 4 12B**
- **Phi-4 14B** / **Qwen3 14B** — possible with Q4, watch for OOM

**Start here if unsure:** open **Model Hub**, search a name, pick a **GGUF** quant (e.g. UD-Q4_K_XL), download — or type `hi` in Chat for a small default.

![Model Hub — search and pick a GGUF](https://raw.githubusercontent.com/StudentOffersco/unsloth-studio-oneclick/main/images/03-model-hub.png)

First download can take a minute; the bar may sit low then jump — that’s normal.

![Model downloading in Studio](https://raw.githubusercontent.com/StudentOffersco/unsloth-studio-oneclick/main/images/02-model-downloading.png)

Skip **20B+ dense** and **70B+** on free Colab T4 — use Kaggle T4×2 or a bigger GPU for those.

### Good to know
- The blank preview inside Colab is normal — use the Cloudflare link / Open button
- Free Colab sessions last up to ~12 hours and may disconnect if idle
- You do **not** need to fine-tune — chatting alone is fine
- This notebook only launches the official Unsloth installer ([`unslothai/unsloth`](https://github.com/unslothai/unsloth)). Studio UI is AGPL-3.0.


```python
# @title ▶ Run this cell — installs and starts Unsloth Studio (~4–5 min first run)
import os, sys, subprocess
from pathlib import Path

ROOT = Path("/content")
REPO = ROOT / "unsloth"
print(f"Working in: {ROOT}")
print("First run usually takes about 4–5 minutes. Please wait...")
print("")

need_clone = not (REPO / "studio" / "setup.sh").exists()
if need_clone:
    if REPO.exists():
        subprocess.run(["rm", "-rf", str(REPO)], check=False)
    print("Downloading Unsloth (first run only)...")
    subprocess.run(
        ["git", "clone", "--depth", "1", "--branch", "main",
         "https://github.com/unslothai/unsloth.git", str(REPO)],
        check=True,
    )
else:
    print("Unsloth already downloaded — reusing it.")

os.chdir(REPO)
subprocess.run(["chmod", "+x", "studio/setup.sh"], check=True)
print("Installing Unsloth Studio...")
r = subprocess.run(["./studio/setup.sh", "--local"])
if r.returncode != 0:
    raise SystemExit(
        "Install failed. Reset the runtime, make sure GPU is on, then run this cell again."
    )

backend = REPO / "studio" / "backend"
if not (backend / "colab.py").exists():
    raise FileNotFoundError("Install finished but Studio files are missing — reset and retry.")

print("")
print("✅ Install done — starting Studio...")
print("")
print("Next steps:")
print("  1) Click Open Unsloth Studio (or open the Cloudflare URL in a new tab)")
print("  2) Enter the password on the card (log in as unsloth)")
print("  3) Model Hub: try Qwen3.5 4B (Q4) — or Chat and type hi")
print("  4) Keep THIS cell running — stopping it shuts Studio down")
print("")

sys.path.insert(0, str(backend))
from colab import start
start()
```

### Stuck?
- **Still installing** — wait the full ~4–5 minutes on first run
- **Blank box inside Colab** — ignore it; use **Open Unsloth Studio** / the Cloudflare URL
- **Password** — use the one on the Shareable Unsloth Link card
- **Forgot the password** — only in the cell output; Runtime → Disconnect and delete runtime → run again for a new one
- **`No module named 'colab'`** — install didn’t finish; reset runtime → GPU on → run again
- **Out of memory** — pick a smaller model (Qwen3.5 4B / Gemma 4 E4B) or a smaller GGUF quant
