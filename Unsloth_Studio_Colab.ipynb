# Unsloth Studio on Google Colab

Run open AI models in your browser for free — **chat** or light **fine-tuning** — using [Unsloth Studio](https://unsloth.ai/docs/new/studio).

### How to use
1. **Runtime → Change runtime type → GPU (T4) → Save**
2. **Run the cell below** (press ▶)

**First run usually takes about 4–5 minutes** (download + install). That’s normal. When it’s done you’ll see a shareable Unsloth card with a Cloudflare link and a password.

### After it starts
1. Click **Open Unsloth Studio** (or copy the Cloudflare URL into a **new tab**)
2. On the login page, enter the **password** shown in the notebook output (the card says to log in as `unsloth` — you mainly need that password)
3. Load a model:
   - **Model Hub** — pick a model, or
   - **Chat** — type `hi` and Studio can download a small default model
4. Leave this Colab cell **running** while you use Studio

### Good to know
- The blank preview inside Colab is normal — use the Cloudflare link / Open button
- Free Colab sessions last up to ~12 hours and may disconnect if idle
- Free T4 GPUs work best with smaller models (roughly under ~22B)
- You do **not** need to fine-tune — chatting alone is fine
- This notebook only launches the official Unsloth installer ([`unslothai/unsloth`](https://github.com/unslothai/unsloth)). Studio UI is AGPL-3.0.


```python
# @title ▶ Run this cell — installs and starts Unsloth Studio (~4–5 min first run)
import os, sys, subprocess
from pathlib import Path

ROOT = Path("/content")
REPO = ROOT / "unsloth"
print(f"Working in: {ROOT}")
print("First run usually takes about 4–5 minutes. Please wait...")
print("")

need_clone = not (REPO / "studio" / "setup.sh").exists()
if need_clone:
    if REPO.exists():
        subprocess.run(["rm", "-rf", str(REPO)], check=False)
    print("Downloading Unsloth (first run only)...")
    subprocess.run(
        ["git", "clone", "--depth", "1", "--branch", "main",
         "https://github.com/unslothai/unsloth.git", str(REPO)],
        check=True,
    )
else:
    print("Unsloth already downloaded — reusing it.")

os.chdir(REPO)
subprocess.run(["chmod", "+x", "studio/setup.sh"], check=True)
print("Installing Unsloth Studio...")
r = subprocess.run(["./studio/setup.sh", "--local"])
if r.returncode != 0:
    raise SystemExit(
        "Install failed. Reset the runtime, make sure GPU is on, then run this cell again."
    )

backend = REPO / "studio" / "backend"
if not (backend / "colab.py").exists():
    raise FileNotFoundError("Install finished but Studio files are missing — reset and retry.")

print("")
print("✅ Install done — starting Studio...")
print("")
print("Next steps:")
print("  1) Click Open Unsloth Studio (or open the Cloudflare URL in a new tab)")
print("  2) Enter the password shown on the card (log in as unsloth)")
print("  3) Model Hub to pick a model — or Chat and type hi for a small default")
print("  4) Keep THIS cell running — stopping it shuts Studio down")
print("")

sys.path.insert(0, str(backend))
from colab import start
start()
```

### Stuck?
- **Still installing** — wait the full ~4–5 minutes on first run
- **Blank box inside Colab** — ignore it; use **Open Unsloth Studio** / the Cloudflare URL
- **Forgot the password** — it’s only in the cell output; Runtime → Disconnect and delete runtime → run again for a new one
- **`No module named 'colab'`** — install didn’t finish; reset runtime → GPU on → run again
